# 1. Video to Coordinates #

```python

from helpers.video_to_coord import process_video_fast

name = "cherrie"
VIDEO_PATH = "./data/cherrie.MOV"
MODEL_PATH = "./pose_landmarker_lite.task"

PROCESS_EVERY_N_FRAMES = 30
PROCESS_FIRST_N_SECONDS = None
SKIP_UNPROCESSED_FRAMES = True


detected_landmarks = process_video_fast(
    video_path=VIDEO_PATH,
    model_path=MODEL_PATH,
    process_every_n_frames=PROCESS_EVERY_N_FRAMES,
    process_first_n_seconds=PROCESS_FIRST_N_SECONDS,
)


from helpers.video_to_coord import landmarks_to_dataframe
df = landmarks_to_dataframe(detected_landmarks)
df
```

# 2. Process Coordinates --> Convert to Angles + Additional Features #

```python
from helpers.calculated_feature import calculate_motion_features_df
# angle1	(11, 13, 15)	Left elbow
# angle2	(12, 14, 16)	Right elbow
# angle3	(11, 13, 23)	Left shoulder
# angle4	(12, 14, 24)	Right shoulder
# angle5	(11, 23, 25)	Left hip
# angle6	(12, 24, 26)	Right hip

feature_df = calculate_motion_features_df(
    df,
    fps=30,
    rom_window_seconds=1.0,
)
feature_df
```

# 3. Add in LDF and AMI#

```python
from helpers.read_ldf import readLDFdata
LDF_bloodflow = readLDFdata("./data/angela")
if len(LDF_bloodflow) < len(feature_df):
    LDF_bloodflow.extend([None] * (len(feature_df) - len(LDF_bloodflow)))
elif len(LDF_bloodflow) > len(feature_df):
    LDF_bloodflow = LDF_bloodflow[:len(feature_df)]
feature_df['LDF_bloodflow'] = LDF_bloodflow
feature_df
```

```python
from helpers.read_ami import readAMIdata
AMI_bloodflow = readAMIdata("./data/angela")
if len(AMI_bloodflow) < len(feature_df):
    AMI_bloodflow.extend([None] * (len(feature_df) - len(AMI_bloodflow)))
elif len(AMI_bloodflow) > len(feature_df):
    AMI_bloodflow = AMI_bloodflow[:len(feature_df)]
feature_df['AMI_bloodflow'] = AMI_bloodflow
feature_df
```

# 4. Save the data

```python
import pandas as pd
feature_df.to_csv("./data/angela_final.csv")
```
