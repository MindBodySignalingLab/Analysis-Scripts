```python
# Read everything in the data file into lines
# lines is a list, each element is one line in the data file
def readLDFdata(filepath):
    file_path = filepath
    file = open("" + file_path + ".txt", "r")
    lines = file.readlines()


    # Finds where the raw data starts
    # Skips the two blank lines at very beginning
    i = 0
    while lines[i] != "4) Trace Data\n":
        i += 1
    i += 2


    # Read All Raw LDF Data
    raw_data_ldf = []
    for j in range(i, len(lines)):
        line = lines[j]
        ldf_val = float(line.split("\t")[4])
        raw_data_ldf.append(ldf_val)

    # See if the last datapoint is correct
    return raw_data_ldf
```

```python
import pandas as pd

LDF_bloodflow = readLDFdata("pose_detection/marco grea eagle spreading wings 05.07.26")
df=pd.read_csv("pose_detection/frame_data_motion_features.csv")
```

```python
every_3_frame = df.iloc[::3]
LDF_bloodflow.extend([None] * (5615 - len(LDF_bloodflow)))
every_3_frame['LDF_bloodflow'] = LDF_bloodflow
final_df = every_3_frame
```

```python
final_df.to_csv("combined_data.csv")
```
